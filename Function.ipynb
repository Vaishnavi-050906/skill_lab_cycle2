```python
def find_avg_marks(marks):
    sum_of_marks=sum(marks)
    no_of_subjects=len(marks)
    avg_marks=sum_of_marks / no_of_subjects
    return avg_marks #positional argument

    def calc_grade(avg_marks):
        if avg_marks >=80:
            grade='A'
        elif avg_marks >=65: 
            grade='B'
        elif avg_marks >=55: 
            grade='C'
        else:
            grade='F'
        return grade


marks=[46,89,60,86,84,93]
avg_marks = find_avg_marks(marks)
print("Avg marks:",avg_marks)
grade = calc_grade(avg_marks)
print("avg grade:",grade)
```
